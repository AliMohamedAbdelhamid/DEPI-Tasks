

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with `median` for numeric and `mode` for categorical

```python
# trying to fill missing values in my datset using for loop 

num_cols = data.select_dtypes("number").columns
str_cols = data.select_dtypes("string").columns

for col in num_cols:
    data[col] = data[col].fillna(data[col].median())

for col in str_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

```python
#check the Nulls 

nulls = data.isnull().sum()
ratio = (nulls / len(data)) * 100
display(pd.DataFrame({'Null Count': nulls, 
                      'Null Ratio (%)': ratio, 
                      'Dtypes': data.dtypes}).sort_values(by='Null Count', ascending=False))
```


# <a id="4">  4 ) . Do the users' ages have outlier values?</a>


> ### <mark>Your analsys and exploration here </mark>

```python
Q1 = data['member_age'].quantile(0.25)
Q3 = data['member_age'].quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print(f"Lower fence: {lower_fence}")
print(f"Upper fence: {upper_fence}")

lower_outliers = data[data['member_age'] < lower_fence]['member_age']
upper_outliers = data[data['member_age'] > upper_fence]['member_age']

print(f"number of lower outliers: {len(lower_outliers)}")
print(f"Lower outliers: {lower_outliers}")

print(f"number of upper outliers: {len(upper_outliers)}")
print(f"Upper outliers: {upper_outliers}")

data['member_age'] = data['member_age'].clip(lower_fence, upper_fence)

plt.figure(figsize=(10, 5), dpi=200)

plt.hist(data['member_age'], bins=20, color='green', edgecolor='black')
plt.title(f'Distribution of users ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
