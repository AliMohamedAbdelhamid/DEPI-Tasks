```python
message = "###!!@emocleW PGTQ!!!6789"
result = ""

# Extract the core part:
for i in message:
    if i.isalpha() or i == " ":
        result += i

print(result)
```

```python
# Reverse the first word:
word = ""

for i in result:
    if i != " ":
        word += i
    else: break

reversed = word[::-1]
print(reversed)
```

```python
# Replace shifted vowels in the second word: 
Sec_part = ""

for i in range(len(result)):
    if result[i] == " ":
        for j in result[i:]:
            Sec_part += j

print(Sec_part)

# Final decoded message:
decoded_message = reversed + Sec_part
print(decoded_message)
```

```python
# Replace shifted vowels in the second word: 
Sec_part2 = ""

for i in range(len(result2)):
    if result2[i] == " ":
        for j in result2[i:]:
            if j == "I":
                j = "E"
            elif j == "O":
                j = "U"
            Sec_part2 += j

print(Sec_part2)
# Final decoded message:

decoded_message2 = reversed2 + Sec_part2
print(decoded_message2)
```

```python
def reverse_words(s):
    s = s.split(" ")
    s = s[::-1]
    result = " ".join(s)
    return result
    

reverse_words("ali mohamed")
```

```python
def are_you_playing_banjo(name):
    if name[0].lower() == "r":
        return name + " plays banjo"
    else:
        return name + " does not play banjo"
```
