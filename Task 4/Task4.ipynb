```python
## 1. Write a function that inputs a number and prints the multiplication table of that number

class func1:

    '''
        Function 1 prints the multiplication table of a number

        Attributes:
            num (int): the number to get its multplication table 
    '''

    def __init__(self, num: int = 0) -> None:
        """
            Set the number of the object

            Parameters:
                num (int): the number to be set in the object
        """
        self.num = num

    def mul_table(self) -> None:
        """
            Prints the multiblication table of a given number

            Parameters:
                num (int): The number to print its multiblication table
            
                Returns:
                    None
        """
        for i in range(1,11):
            result = self.num * i
            print(f"{self.num} * {i} = {result}")


c1 = func1(8)
c1.mul_table()
```

```python
## 2. Write a program to print twin primes less than 1000. 
## If two consecutive odd numbers are both prime then they are known as 
## twin primes 

class func2:

    '''
        Function 2 extract the consecutive odd prime numbers in specified range and print the result

        Attributes:
            None
    '''

    def __init__(self, num: int = 1) -> None:
        self.num = num
        
    def is_prime(self, num:int) -> bool:

        """ 
            Determines if the number is prime or not

            Parameters:
                num (int): The number to determine if it's a prime number of not
            Returns:
                prime (bool)
        """
        prime = True
        if num == 1: 
            prime = False
            return prime

        for i in range(2,num):
            if num % i == 0:
                prime = False
                break

        return prime

    def is_twin_primes(self) -> None:
        """
            Prints the twin primes between 1:1000

            Parameters:
                None
            Returns:
                None
        """
        print("Twin Prime:")
        for i in range(1,1000,2):
            if self.is_prime(i) and self.is_prime(i+2):
                print(f"{i} and {i+2}")

d1 = func2(5)
print(d1.num)
d1.is_twin_primes()
```

```python
## 3. Write a program to find out the prime factors of a number. 
## Example: prime factors of 56 - 2, 2, 2, 7


class func3:
    """
        Function 3 prints the prime factors of a number
    """

    def __init__(self, num: int) -> None:
        self.num = num
        
    def is_prime(self, num:int) -> bool:

        """ 
            Determines if the number is prime or not

            Parameters:
                num (int): The number to determine if it's a prime number of not
            Returns:
                prime (bool)
        """
        prime = True
        if num == 1: 
            prime = False
            return prime

        for i in range(2,num):
            if num % i == 0:
                prime = False
                break

        return prime


    def prime_factors(self, num: int) -> list:

        """
            Determines the prime factors of a given number

            Parameters:
                num (int): The number to get its prime factors
            Returns:
                factor (list)
            
        """
        factors = []

        for i in range(2,num):

            if self.is_prime(num):
                factors.append(num) 
                return factors

            elif num % i == 0:
                remain = int(num / i)
                factors.append(i)
                # print("before if-else condition")
                # print(f"now num = {num} and factors is {factors} and remain is {remain}")

                if not self.is_prime(remain):
                    # print(f"{remain} is not prime")
                    factors.extend(self.prime_factors(remain))
                    # print("after if condition")
                    # print(f"now num = {num} and factors is {factors} and remain is {remain}")
                else:
                    # print(f"{remain} is prime")
                    factors.append(remain)
                    # print("after else condition")
                    # print(f"now num = {num} and factors is {factors} and remain is {remain}")
                return factors

s1 = func3(3)
print(s1.num)

print(s1.prime_factors(56))
print(s1.prime_factors(12))
```

```python
## 4.  Write a function that converts a decimal number to binary number 


class func4:

    """
        Function 4 converts a decimal number to binary number
    """

    def __init__(self, num:int) -> None:
        self.num = num
        
    def dec_2_bin(self) -> None:

        """
            Converts a decimal number to binary number

            Parameters:
                num (int): The number to be changed to binary
            Returns:
                None
        """

        finish = False
        remaining = self.num
        binary = ""

        while not finish:    
            result = remaining % 2
            if result != 0:
                binary += "1"
            else:
                binary += "0"
            remaining = remaining // 2
            
            if remaining == 0:
                finish = True

        binary = binary[::-1]
        print(binary)


h1 = func4(11)
h1.dec_2_bin()
        
        
```

```python
## 5. A number is called perfect if the sum of proper divisors of that number is equal to the number. 
## For example 28 is perfect number, since 1+2+4+7+14=28. 
## Write a program to print all the perfect numbers in a given range 


class func5:

    '''
        Function 5 extract the perfect number in specified range
    '''

    def __init__(self, num1: int, num2: int) -> None:
        self.start = num1
        self.end = num2


    def is_prime(self, num:int) -> bool:

        """ 
            Determines if the number is prime or not

            Parameters:
                num (int): The number to determine if it's a prime number of not
            Returns:
                prime (bool)
        """
        prime = True
        if num == 1: 
            prime = False
            return prime

        for i in range(2,num):
            if num % i == 0:
                prime = False
                break

        return prime

      
   
    def is_perfect_num(self) -> None:

        """
            Prints the perfect numebrs in a given range and there corrisponding sum of divisors

            Parameters:
                num1 (int): Range start
                num2 (int): Range end

            Returns:
                None
        """
        
        divisors = []
        sum = 0

        for i in range(self.start,self.end+1):
            # print(f"we now in {i}")
            if self.is_prime(i):
                divisors.append(1)
            else:
                for j in range(1,i):
                    if i % j == 0:
                        # global sum
                        sum += j
                        divisors.append(j)
                # print(f"The sum = {sum}")
                if sum == i:
                    print(i, end=": \n")
                    print(*divisors, sep = " + ", end = f" = {sum}\n")
            # print(divisors)
            divisors.clear()
            sum = 0

g1 = func5(0,100)
g1.is_perfect_num()
```
